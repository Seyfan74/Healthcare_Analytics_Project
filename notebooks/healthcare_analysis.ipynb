```python
# ============================================================
# HEALTHCARE EDA VALIDATION
# ============================================================

def validate_eda(data, eda_results):

    print("=" * 70)
    print("HEALTHCARE EDA VALIDATION")
    print("=" * 70)

    validation_results = {}

    # --------------------------------------------------------
    # EXPECTED SECTIONS
    # --------------------------------------------------------

    expected_sections = [
        "patients",
        "admissions",
        "departments",
        "doctors",
        "diagnoses",
        "diagnosis_records",
        "lab_tests",
        "lab_results",
        "medications",
        "medication_records",
        "procedures",
        "procedure_records",
        "insurance_claims",
        "payments"
    ]

    # --------------------------------------------------------
    # SECTION VALIDATION
    # --------------------------------------------------------

    print("\nSECTION VALIDATION")
    print("-" * 70)

    for section in expected_sections:

        if section in eda_results:
            print(f"✓ Section exists: {section}")
            validation_results[section] = {
                "section_exists": True
            }

        else:
            print(f"✗ Missing section: {section}")
            validation_results[section] = {
                "section_exists": False
            }

    # --------------------------------------------------------
    # RECORD COUNT VALIDATION
    # --------------------------------------------------------

    print("\n" + "-" * 70)
    print("RECORD COUNT VALIDATION")
    print("-" * 70)

    # The actual record-count key used by each EDA section.
    #
    # IMPORTANT:
    # We use only keys that are expected to represent
    # the number of rows/records in each table.
    # --------------------------------------------------------

    record_mapping = {
    "patients": "total_patients",
    "admissions": "total_admissions",
    "departments": "total_departments",
    "doctors": "total_doctors",
    "diagnoses": "total_diagnoses",
    "diagnosis_records": "total_records",
    "lab_tests": "total_lab_tests",
    "lab_results": "total_lab_results",
    "medications": "total_medications",
    "medication_records": "total_records",
    "procedures": "total_procedures",
    "procedure_records": "total_records",
    "insurance_claims": "record_count",
    "payments": "total_payments"
}

    for table, result_key in record_mapping.items():

        # Check that the table exists in data
        if table not in data:
            print(f"✗ {table}: table missing from data")
            continue

        # Check that the EDA section exists
        if table not in eda_results:
            print(f"✗ {table}: EDA section missing")
            continue

        # Check that the expected EDA metric exists
        if result_key not in eda_results[table]:

            print(
                f"⚠ {table}: EDA key '{result_key}' "
                f"not found"
            )

            print(
                "  Available keys:",
                list(eda_results[table].keys())
            )

            continue

        actual_count = len(data[table])
        eda_count = eda_results[table][result_key]

        if actual_count == eda_count:

            print(
                f"✓ {table}: "
                f"EDA={eda_count}, Actual={actual_count}"
            )

            validation_results[table]["record_count"] = {
                "eda": eda_count,
                "actual": actual_count,
                "match": True
            }

        else:

            print(
                f"✗ {table}: "
                f"EDA={eda_count}, Actual={actual_count}"
            )

            validation_results[table]["record_count"] = {
                "eda": eda_count,
                "actual": actual_count,
                "match": False
            }

    # --------------------------------------------------------
    # SUMMARY
    # --------------------------------------------------------

    print("\n" + "=" * 70)
    print("VALIDATION COMPLETE")
    print("=" * 70)

    return validation_results
```

```python

# ============================================================
# FINAL EDA VALIDATION FUNCTION
# ============================================================

def validate_eda(data, eda_results):

    print("=" * 70)
    print("EDA VALIDATION")
    print("=" * 70)

    # --------------------------------------------------------
    # Expected EDA result keys and actual DataFrame tables
    # --------------------------------------------------------
    record_mapping = {
    "patients": "total_patients",
    "admissions": "total_admissions",
    "departments": "total_departments",
    "diagnoses": "total_diagnoses",
    "diagnosis_records": "total_records",
    "doctors": "total_doctors",
    "insurance_claims": "record_count",
    "lab_results": "total_lab_results",
    "lab_tests": "total_lab_tests",
    "medication_records": "total_records",
    "medications": "total_medications",
    "payments": "total_payments",
    "procedure_records": "total_records",
    "procedures": "total_procedures"
}

    validation_results = {}

    # --------------------------------------------------------
    # Validate record counts
    # --------------------------------------------------------
    for table, result_key in record_mapping.items():

        print("\n" + "-" * 70)
        print(table.upper())

        # Check that table exists in data
        if table not in data:
            print("❌ Table not found in data")
            validation_results[table] = False
            continue

        # Check that table exists in EDA results
        if table not in eda_results:
            print("❌ Table not found in EDA results")
            validation_results[table] = False
            continue

        # Check that expected EDA key exists
        if result_key not in eda_results[table]:
            print(f"❌ EDA key '{result_key}' not found")
            print("Available keys:")
            print(list(eda_results[table].keys()))
            validation_results[table] = False
            continue

        # Actual number of rows
        actual_count = len(data[table])

        # EDA value
        eda_count = eda_results[table][result_key]

        print("Expected / Actual DataFrame count:", actual_count)
        print("EDA count:", eda_count)

        # Compare
        if actual_count == eda_count:
            print("✅ PASS")
            validation_results[table] = True
        else:
            print("❌ FAIL")
            validation_results[table] = False

    # --------------------------------------------------------
    # Patient ID validation
    # --------------------------------------------------------
    print("\n" + "=" * 70)
    print("PATIENT ID VALIDATION")
    print("=" * 70)

    if "patients" in data and "patients" in eda_results:

        actual_unique_patients = data["patients"]["patient_id"].nunique()

        eda_unique_patients = eda_results["patients"].get(
            "unique_patients"
        )

        print("Actual unique patient IDs:", actual_unique_patients)
        print("EDA unique patient IDs:", eda_unique_patients)

        if actual_unique_patients == eda_unique_patients:
            print("✅ Unique patient IDs PASS")
        else:
            print("❌ Unique patient IDs FAIL")

    # --------------------------------------------------------
    # Summary
    # --------------------------------------------------------
    print("\n" + "=" * 70)
    print("FINAL VALIDATION SUMMARY")
    print("=" * 70)

    passed = sum(validation_results.values())
    total = len(validation_results)

    print(f"Passed: {passed}/{total}")
    print(f"Failed: {total - passed}/{total}")

    if passed == total:
        print("\n🎉 ALL EDA RECORD COUNT VALIDATIONS PASSED!")
    else:
        print("\n⚠️ SOME EDA VALIDATIONS FAILED.")
        print("Check the tables marked ❌ above.")

    return validation_results

```

```python
# ============================================================
# HEALTHCARE EDA METRIC VALIDATION
# ============================================================

import pandas as pd
import numpy as np


def validate_eda_metrics(data, eda_results):

    print("=" * 70)
    print("EDA METRIC VALIDATION")
    print("=" * 70)

    validation_results = {}

    passed = 0
    failed = 0

    def check_metric(section, metric, actual, eda_value, tolerance=0.01):
        """
        Compare an independently calculated value with the EDA value.
        """

        nonlocal passed, failed

        # Convert numeric values safely
        try:
            actual_num = float(actual)
            eda_num = float(eda_value)

            match = np.isclose(
                actual_num,
                eda_num,
                atol=tolerance,
                rtol=1e-6
            )

        except (TypeError, ValueError):
            match = actual == eda_value

        if match:
            print(
                f"  {metric}: "
                f"EDA={eda_value}, Actual={actual} "
                f"✅ PASS"
            )
            passed += 1
        else:
            print(
                f"  {metric}: "
                f"EDA={eda_value}, Actual={actual} "
                f"❌ FAIL"
            )
            failed += 1

        validation_results.setdefault(section, {})[metric] = {
            "eda": eda_value,
            "actual": actual,
            "match": bool(match)
        }

    # ========================================================
    # PATIENTS
    # ========================================================

    print("\n" + "-" * 70)
    print("PATIENTS")
    print("-" * 70)

    patients = data["patients"]

    check_metric(
        "patients",
        "total_patients",
        len(patients),
        eda_results["patients"]["total_patients"]
    )

    check_metric(
        "patients",
        "unique_patients",
        patients["patient_id"].nunique(),
        eda_results["patients"]["unique_patients"]
    )

    check_metric(
        "patients",
        "duplicate_patient_rows",
        patients["patient_id"].duplicated().sum(),
        eda_results["patients"]["duplicate_patient_rows"]
    )

    # ========================================================
    # ADMISSIONS
    # ========================================================

    print("\n" + "-" * 70)
    print("ADMISSIONS")
    print("-" * 70)

    admissions = data["admissions"]

    check_metric(
        "admissions",
        "total_admissions",
        len(admissions),
        eda_results["admissions"]["total_admissions"]
    )

    check_metric(
        "admissions",
        "unique_admissions",
        admissions["admission_id"].nunique(),
        eda_results["admissions"]["unique_admissions"]
    )

    check_metric(
        "admissions",
        "unique_patients",
        admissions["patient_id"].nunique(),
        eda_results["admissions"]["unique_patients"]
    )

    check_metric(
        "admissions",
        "unique_doctors",
        admissions["doctor_id"].nunique(),
        eda_results["admissions"]["unique_doctors"]
    )

    check_metric(
        "admissions",
        "unique_departments",
        admissions["department_id"].nunique(),
        eda_results["admissions"]["unique_departments"]
    )

    # Length of stay
    los = pd.to_numeric(
        admissions["length_of_stay"],
        errors="coerce"
    )

    check_metric(
        "admissions",
        "average_length_of_stay",
        los.mean(),
        eda_results["admissions"]["average_length_of_stay"]
    )

    check_metric(
        "admissions",
        "median_length_of_stay",
        los.median(),
        eda_results["admissions"]["median_length_of_stay"]
    )

    check_metric(
        "admissions",
        "maximum_length_of_stay",
        los.max(),
        eda_results["admissions"]["maximum_length_of_stay"]
    )

    # Patient satisfaction
    satisfaction = pd.to_numeric(
        admissions["patient_satisfaction"],
        errors="coerce"
    )

    check_metric(
        "admissions",
        "average_patient_satisfaction",
        satisfaction.mean(),
        eda_results["admissions"]["average_patient_satisfaction"]
    )

    check_metric(
        "admissions",
        "median_patient_satisfaction",
        satisfaction.median(),
        eda_results["admissions"]["median_patient_satisfaction"]
    )

    # Readmission rate
    readmission = pd.to_numeric(
        admissions["readmission_30_days"],
        errors="coerce"
    )

    actual_readmission_rate = (
        readmission.mean() * 100
    )

    check_metric(
        "admissions",
        "readmission_rate",
        actual_readmission_rate,
        eda_results["admissions"]["readmission_rate"]
    )

    # ========================================================
    # DEPARTMENTS
    # ========================================================

    print("\n" + "-" * 70)
    print("DEPARTMENTS")
    print("-" * 70)

    departments = data["departments"]

    check_metric(
        "departments",
        "total_departments",
        len(departments),
        eda_results["departments"]["total_departments"]
    )

    check_metric(
        "departments",
        "unique_departments",
        departments["department_id"].nunique(),
        eda_results["departments"]["unique_departments"]
    )

    # ========================================================
    # DOCTORS
    # ========================================================

    print("\n" + "-" * 70)
    print("DOCTORS")
    print("-" * 70)

    doctors = data["doctors"]

    check_metric(
        "doctors",
        "total_doctors",
        len(doctors),
        eda_results["doctors"]["total_doctors"]
    )

    check_metric(
        "doctors",
        "unique_doctors",
        doctors["doctor_id"].nunique(),
        eda_results["doctors"]["unique_doctors"]
    )

    # ========================================================
    # DIAGNOSES
    # ========================================================

    print("\n" + "-" * 70)
    print("DIAGNOSES")
    print("-" * 70)

    diagnoses = data["diagnoses"]

    check_metric(
        "diagnoses",
        "total_diagnoses",
        len(diagnoses),
        eda_results["diagnoses"]["total_diagnoses"]
    )

    check_metric(
        "diagnoses",
        "unique_diagnoses",
        diagnoses["diagnosis_id"].nunique(),
        eda_results["diagnoses"]["unique_diagnoses"]
    )

    # ========================================================
    # DIAGNOSIS RECORDS
    # ========================================================

    print("\n" + "-" * 70)
    print("DIAGNOSIS RECORDS")
    print("-" * 70)

    diagnosis_records = data["diagnosis_records"]

    check_metric(
        "diagnosis_records",
        "total_records",
        len(diagnosis_records),
        eda_results["diagnosis_records"]["total_records"]
    )

    check_metric(
        "diagnosis_records",
        "unique_records",
        diagnosis_records["diagnosis_record_id"].nunique(),
        eda_results["diagnosis_records"]["unique_records"]
    )

    check_metric(
        "diagnosis_records",
        "unique_admissions",
        diagnosis_records["admission_id"].nunique(),
        eda_results["diagnosis_records"]["unique_admissions"]
    )

    # ========================================================
    # LAB TESTS
    # ========================================================

    print("\n" + "-" * 70)
    print("LAB TESTS")
    print("-" * 70)

    lab_tests = data["lab_tests"]

    check_metric(
        "lab_tests",
        "total_lab_tests",
        len(lab_tests),
        eda_results["lab_tests"]["total_lab_tests"]
    )

    check_metric(
        "lab_tests",
        "unique_lab_tests",
        lab_tests["lab_test_id"].nunique(),
        eda_results["lab_tests"]["unique_lab_tests"]
    )

    # ========================================================
    # LAB RESULTS
    # ========================================================

    print("\n" + "-" * 70)
    print("LAB RESULTS")
    print("-" * 70)

    lab_results = data["lab_results"]

    check_metric(
        "lab_results",
        "total_lab_results",
        len(lab_results),
        eda_results["lab_results"]["total_lab_results"]
    )

    check_metric(
        "lab_results",
        "unique_lab_results",
        lab_results["lab_result_id"].nunique(),
        eda_results["lab_results"]["unique_lab_results"]
    )

    check_metric(
        "lab_results",
        "unique_admissions",
        lab_results["admission_id"].nunique(),
        eda_results["lab_results"]["unique_admissions"]
    )

    check_metric(
        "lab_results",
        "unique_lab_tests",
        lab_results["lab_test_id"].nunique(),
        eda_results["lab_results"]["unique_lab_tests"]
    )

    result_value = pd.to_numeric(
        lab_results["result_value"],
        errors="coerce"
    )

    check_metric(
        "lab_results",
        "average_result_value",
        result_value.mean(),
        eda_results["lab_results"]["average_result_value"]
    )

    check_metric(
        "lab_results",
        "median_result_value",
        result_value.median(),
        eda_results["lab_results"]["median_result_value"]
    )

    # ========================================================
    # MEDICATIONS
    # ========================================================

    print("\n" + "-" * 70)
    print("MEDICATIONS")
    print("-" * 70)

    medications = data["medications"]

    check_metric(
        "medications",
        "total_medications",
        len(medications),
        eda_results["medications"]["total_medications"]
    )

    check_metric(
        "medications",
        "unique_medications",
        medications["medication_id"].nunique(),
        eda_results["medications"]["unique_medications"]
    )

    # ========================================================
    # MEDICATION RECORDS
    # ========================================================

    print("\n" + "-" * 70)
    print("MEDICATION RECORDS")
    print("-" * 70)

    medication_records = data["medication_records"]

    check_metric(
        "medication_records",
        "total_records",
        len(medication_records),
        eda_results["medication_records"]["total_records"]
    )

    check_metric(
        "medication_records",
        "unique_records",
        medication_records["medication_record_id"].nunique(),
        eda_results["medication_records"]["unique_records"]
    )

    check_metric(
        "medication_records",
        "unique_admissions",
        medication_records["admission_id"].nunique(),
        eda_results["medication_records"]["unique_admissions"]
    )

    # ========================================================
    # PROCEDURES
    # ========================================================

    print("\n" + "-" * 70)
    print("PROCEDURES")
    print("-" * 70)

    procedures = data["procedures"]

    check_metric(
        "procedures",
        "total_procedures",
        len(procedures),
        eda_results["procedures"]["total_procedures"]
    )

    check_metric(
        "procedures",
        "unique_procedures",
        procedures["procedure_id"].nunique(),
        eda_results["procedures"]["unique_procedures"]
    )

    # ========================================================
    # PROCEDURE RECORDS
    # ========================================================

    print("\n" + "-" * 70)
    print("PROCEDURE RECORDS")
    print("-" * 70)

    procedure_records = data["procedure_records"]

    check_metric(
        "procedure_records",
        "total_records",
        len(procedure_records),
        eda_results["procedure_records"]["total_records"]
    )

    check_metric(
        "procedure_records",
        "unique_records",
        procedure_records["procedure_record_id"].nunique(),
        eda_results["procedure_records"]["unique_records"]
    )

    check_metric(
        "procedure_records",
        "unique_admissions",
        procedure_records["admission_id"].nunique(),
        eda_results["procedure_records"]["unique_admissions"]
    )

    check_metric(
        "procedure_records",
        "unique_procedures",
        procedure_records["procedure_id"].nunique(),
        eda_results["procedure_records"]["unique_procedures"]
    )

    actual_total_cost = pd.to_numeric(
        procedure_records["actual_cost"],
        errors="coerce"
    ).sum()

    actual_average_cost = pd.to_numeric(
        procedure_records["actual_cost"],
        errors="coerce"
    ).mean()

    actual_median_cost = pd.to_numeric(
        procedure_records["actual_cost"],
        errors="coerce"
    ).median()

    check_metric(
        "procedure_records",
        "total_actual_cost",
        actual_total_cost,
        eda_results["procedure_records"]["total_actual_cost"]
    )

    check_metric(
        "procedure_records",
        "average_actual_cost",
        actual_average_cost,
        eda_results["procedure_records"]["average_actual_cost"]
    )

    check_metric(
        "procedure_records",
        "median_actual_cost",
        actual_median_cost,
        eda_results["procedure_records"]["median_actual_cost"]
    )

    # ========================================================
    # INSURANCE CLAIMS
    # ========================================================

    print("\n" + "-" * 70)
    print("INSURANCE CLAIMS")
    print("-" * 70)

    claims = data["insurance_claims"]

    check_metric(
        "insurance_claims",
        "record_count",
        len(claims),
        eda_results["insurance_claims"]["record_count"]
    )

    billed = pd.to_numeric(
        claims["billed_amount"],
        errors="coerce"
    )

    approved = pd.to_numeric(
        claims["approved_amount"],
        errors="coerce"
    )

    denied = pd.to_numeric(
        claims["denied_amount"],
        errors="coerce"
    )

    actual_total_billed = billed.sum()
    actual_total_approved = approved.sum()
    actual_total_denied = denied.sum()

    check_metric(
        "insurance_claims",
        "total_billed",
        actual_total_billed,
        eda_results["insurance_claims"]["total_billed"]
    )

    check_metric(
        "insurance_claims",
        "total_approved",
        actual_total_approved,
        eda_results["insurance_claims"]["total_approved"]
    )

    check_metric(
        "insurance_claims",
        "total_denied",
        actual_total_denied,
        eda_results["insurance_claims"]["total_denied"]
    )

    actual_approval_rate = (
        actual_total_approved / actual_total_billed
    ) * 100

    actual_denial_rate = (
        actual_total_denied / actual_total_billed
    ) * 100

    check_metric(
        "insurance_claims",
        "approval_rate",
        actual_approval_rate,
        eda_results["insurance_claims"]["approval_rate"]
    )

    check_metric(
        "insurance_claims",
        "denial_rate",
        actual_denial_rate,
        eda_results["insurance_claims"]["denial_rate"]
    )

    # ========================================================
    # PAYMENTS
    # ========================================================

    print("\n" + "-" * 70)
    print("PAYMENTS")
    print("-" * 70)

    payments = data["payments"]

    check_metric(
        "payments",
        "total_payments",
        len(payments),
        eda_results["payments"]["total_payments"]
    )

    check_metric(
        "payments",
        "unique_payments",
        payments["payment_id"].nunique(),
        eda_results["payments"]["unique_payments"]
    )

    check_metric(
        "payments",
        "unique_claims",
        payments["claim_id"].nunique(),
        eda_results["payments"]["unique_claims"]
    )

    check_metric(
        "payments",
        "unique_patients",
        payments["patient_id"].nunique(),
        eda_results["payments"]["unique_patients"]
    )

    payment_amount = pd.to_numeric(
        payments["payment_amount"],
        errors="coerce"
    )

    check_metric(
        "payments",
        "total_payment_amount",
        payment_amount.sum(),
        eda_results["payments"]["total_payment_amount"]
    )

    check_metric(
        "payments",
        "average_payment_amount",
        payment_amount.mean(),
        eda_results["payments"]["average_payment_amount"]
    )

    check_metric(
        "payments",
        "median_payment_amount",
        payment_amount.median(),
        eda_results["payments"]["median_payment_amount"]
    )

    # ========================================================
    # FINAL SUMMARY
    # ========================================================

    print("\n" + "=" * 70)
    print("FINAL METRIC VALIDATION SUMMARY")
    print("=" * 70)

    total = passed + failed

    print(f"Passed: {passed}/{total}")
    print(f"Failed: {failed}/{total}")

    if failed == 0:
        print("\n🎉 ALL EDA METRIC VALIDATIONS PASSED!")
    else:
        print("\n⚠️ SOME EDA METRIC VALIDATIONS FAILED.")
        print("Review the metrics marked ❌ above.")

    return validation_results
```
